***Simulating a Memory Loop***

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage

# We use .strip() to ensure no hidden spaces cause a 400 error
MODEL_NAME = "qwen3.5:9b".strip()

llm_test = ChatOllama(model=MODEL_NAME, temperature=0)

try:
    print(f"Connecting to: '{MODEL_NAME}'...")
    # Passing a list of one message
    response = llm_test.invoke([HumanMessage(content="Hi")])
    print("Success! Response:", response.content)
except Exception as e:
    print(f"Still failing. Error details: {e}")
```

```python
# 1. The Script starts with a System Role
messages = [
    SystemMessage(content="You are SQLmind, a disaster management expert. Be very concise.")
]

# 2. First Human Question
messages.append(HumanMessage(content="Is the sensor at North Hillside active?"))

# 3. Get the first AI response and ADD IT to the script
ai_response = llm.invoke(messages)
messages.append(ai_response) # This is the "Memory" step!
print("First AI Response:", ai_response.content)

# 4. The Pivot (Follow-up question)
# Note: We don't say "North Hillside" again. We just say "it".
messages.append(HumanMessage(content="What about the one at East Valley?"))

# 5. Get the final response with the FULL history
final_response = llm.invoke(messages)
print("\nFollow-up Response:", final_response.content)
```

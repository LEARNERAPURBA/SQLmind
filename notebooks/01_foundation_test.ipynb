```python
from langchain_ollama import ChatOllama
```

```python
# Initialize the SQLmind brain
llm = ChatOllama(
    model="qwen3.5:9b",
    temperature=0 
)

# The 'Ping' Test
try:
    print("Pinging local Qwen model...")
    response = llm.invoke("Hello! Are you ready to act as the brain for SQLmind?")
    print("\n--- Response Received ---")
    print(response)
except Exception as e:
    print(f"\nConnection Failed. Ensure Ollama is running! Error: {e}")
```
